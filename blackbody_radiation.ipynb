```python
import numpy as np
from math import e
import matplotlib.pyplot as plt
```

```python
C1 = 3.742e8 # first radiation constant, 2(pi)hc
C2 = 1.439e4 # second radiation constant, hc/k
```

```python
def blackbody_radiation(λ, T):        # calculates the blackbody spectral radiant emittance 
    E = C1/(λ**5*(e**(C2/(λ*T))-1))   # at wavelength λ and temperature T
    return E
```

```python
wavelengths = np.linspace(0,1,101) # in micrometers
T_sun = 5800 # K
```

```python
plt.figure()
plt.plot(wavelengths, blackbody_radiation(wavelengths, T_sun))
plt.title("Blackbody Radiation Emitted by the Sun", fontsize = 15)
plt.xlabel("λ (μm)", fontsize = 15)
plt.ylabel("$E_{λ,b}$ (W/$m^2$μm)", fontsize = 15)
plt.show()
```
